```python
"""
The goal is to bridge Computer Vision and Natural Language Processing into a single multimodal neural network. The model is given a raw grayscale clothing image and must automatically describe the item in complete English sentences, generating one word at a time based on what it sees in the image and what words it has already spoken.

---

### The Data and Caption Formulation

1. **The Dataset Used**:
   Instead of using massive photo datasets like Flickr8k or MS-COCO (which require gigabytes of storage and hours of GPU time), the experiment uses the built-in **Fashion-MNIST** dataset. It contains $28 \times 28$ grayscale images across 10 clothing categories, such as t-shirt, trouser, pullover, dress, coat, sandal, shirt, sneaker, bag, and ankle boot.

2. **Creating the Training Set**:
   To keep CPU execution fast and practical for a lab session, the notebook samples a balanced subset: **4,000 images for training** and **500 images for testing**. All pixel intensities are normalized into the range of 0 to 1 and expanded to include a single color channel dimension.

3. **Generating Natural Language Sentences**:
   To turn image classification into a genuine language generation task, the code pairs each image with a varied English sentence using 5 randomized sentence templates, such as:
   - *"a photo of a bag"*
   - *"this picture shows a trouser"*
   - *"an image containing a sneaker"*
   - *"a small picture of a coat"*
   - *"here is a dress"*
   
   Every caption is wrapped with two special control markers: `startseq` at the beginning (which signals the model to start generating text) and `endseq` at the end (which tells the model when the sentence is finished).

---

### Text Processing and Sequence Expansion

1. **Tokenization and Vocabulary**:
   A Keras Tokenizer scans the 4,000 generated training captions and assigns a unique integer ID to each unique word. This results in a vocabulary of **27 unique tokens** (including special tokens and index 0 for zero-padding). The longest caption spans **9 words**.

2. **Step-by-Step Training Pairs (Teacher Forcing)**:
   A neural network cannot learn an entire sentence in one shot; it must learn to predict the next word given the image and the partial sentence already spoken. 
   
   For example, for the caption *"startseq a dress endseq"*, the algorithm breaks it into three training pairs:
   - Given the image and `[startseq]`, predict `"a"`.
   - Given the image and `[startseq, a]`, predict `"dress"`.
   - Given the image and `[startseq, a, dress]`, predict `"endseq"`.
   
   All partial sequences are zero-padded to a uniform length of 9 tokens. Expanding all 4,000 training images produces **24,819 individual training pairs**.

---

### The Dual-Branch "Merge" Architecture

The network uses the Keras Functional API to build a two-pathway architecture:

- **The Visual Pathway (CNN Image Encoder)**:
  Acts as the "eyes" of the system. It passes the $28 \times 28$ image through two convolutional blocks (32 filters followed by 64 filters, each with $3 \times 3$ kernels and max-pooling). A global average pooling layer compresses the spatial maps, and a fully connected layer with dropout projects the visual cues into a dense 128-dimensional vector representing the image's content.

- **The Language Pathway (LSTM Caption Decoder)**:
  Acts as the "memory" of the system. It accepts the sequence of words generated so far, maps each word into a 128-dimensional embedding vector, and feeds them into an LSTM layer with 128 memory units. The LSTM keeps track of grammar, phrasing, and the order of previously generated words.

- **The Merging and Prediction Head**:
  The 128-dimensional image vector and the 128-dimensional text vector are combined using an addition layer. This fused representation passes through another dense layer of 128 neurons (ReLU activation) and finally through a softmax output layer of 27 neurons. The output is a probability distribution over all 27 words in the vocabulary, choosing the word with the highest probability.

In total, the model contains **182,171 trainable parameters** (about 711 KB in memory), with the LSTM layer being the largest component (131,584 parameters).

---

### Training Progression and Results

The network was trained on the 24,819 sequence pairs using the Adam optimizer and categorical cross-entropy loss with a batch size of 64 over 10 epochs:

- **Rapid Initial Learning**:
  In Epoch 1, the model started with a training accuracy of 62.69% and a loss of 1.1169, while achieving a validation accuracy of 73.69%. 
- **Steady Convergence**:
  Across epochs 2 through 5, the model quickly picked up the grammatical structure of the sentence templates. By Epoch 5, training accuracy climbed to 79.37% and validation accuracy broke past 80%.
- **Final Epochs (Epochs 6 to 10)**:
  The model continued to refine its predictions smoothly without severe overfitting. Training accuracy peaked at **81.09%** with a training loss of **0.4083**. Validation accuracy reached its peak at **81.67%** at Epoch 9, and concluded at **80.82%** with a validation loss of 0.4200 at Epoch 10.
- Each epoch took approximately 10 to 11 seconds to run on CPU.

---

### How Caption Generation Works at Test Time and Final Outputs

1. **Greedy Word-by-Word Decoding**:
   When given an unseen test image, the trained model generates a caption step-by-step:
   - It starts with the initial prompt `"startseq"`.
   - It feeds the test image and the current prompt into the network to predict the highest-probability next word.
   - It appends that predicted word to the sequence and repeats the process.
   - As soon as the network outputs `"endseq"`, the loop stops immediately, trims off the markers, and returns the final sentence.

2. **The Final Output Produced**:
   - The notebook successfully tested this pipeline on the first test sample and printed:
     `Example generated caption: a photo of a t shirt`
   - The caption perfectly matched the item in the image while maintaining natural English grammar.

3. **Saved Files on Disk**:
   - **`caption_training_loss.png`**: A line plot showing how training loss and validation loss steadily dropped in tandem over the 10 epochs.
   - **`caption_predictions.png`**: A 6-image grid comparing unseen test clothing items, displaying the true item label alongside the complete sentence generated by the model for each.
"""

import random
import numpy as np
import matplotlib.pyplot as plt
from tensorflow.keras.datasets import fashion_mnist
from tensorflow.keras.preprocessing.text import Tokenizer
from tensorflow.keras.preprocessing.sequence import pad_sequences
from tensorflow.keras.utils import to_categorical
from tensorflow.keras.models import Model
from tensorflow.keras.layers import (
    Input, Conv2D, MaxPooling2D, GlobalAveragePooling2D, Dense,
    Embedding, LSTM, Add, Dropout
)

SEED = 42
random.seed(SEED)
np.random.seed(SEED)

# -------------------------------------------------------------------
# 1. Build (image, caption) pairs from Fashion-MNIST
# -------------------------------------------------------------------
(x_train, y_train), (x_test, y_test) = fashion_mnist.load_data()

# Normalize to [0,1] and add a channel dimension: (N, 28, 28) -> (N, 28, 28, 1)
x_train = x_train.astype("float32")[..., np.newaxis] / 255.0
x_test = x_test.astype("float32")[..., np.newaxis] / 255.0

# Simplified class names (avoiding slashes/punctuation, which is friendlier
# for a word-level tokenizer than the official "T-shirt/top" label).
class_names = ["t-shirt", "trouser", "pullover", "dress", "coat",
               "sandal", "shirt", "sneaker", "bag", "ankle boot"]

# A few caption templates per class -> gives linguistic variety instead
# of one fixed sentence per class, which is what makes this a genuine
# sequence-generation problem rather than pure classification.
TEMPLATES = [
    "a photo of a {}",
    "this picture shows a {}",
    "an image containing a {}",
    "a small picture of a {}",
    "here is a {}",
]


def make_caption(label_idx):
    name = class_names[label_idx]
    article_fix = "an " + name if name[0] in "aeiou" else "a " + name
    template = random.choice(TEMPLATES)
    return "startseq " + template.format(name).replace("a " + name, article_fix) + " endseq"


# To keep training fast for a lab session, use a subset of Fashion-MNIST.
N_TRAIN, N_TEST = 4000, 500
train_idx = np.random.choice(len(x_train), N_TRAIN, replace=False)
test_idx = np.random.choice(len(x_test), N_TEST, replace=False)

train_images = x_train[train_idx]
train_captions = [make_caption(int(y_train[i])) for i in train_idx]
test_images = x_test[test_idx]
test_captions = [make_caption(int(y_test[i])) for i in test_idx]

print("Example captions:", train_captions[:3])

# -------------------------------------------------------------------
# 2. Tokenize captions and build the vocabulary
# -------------------------------------------------------------------
tokenizer = Tokenizer(oov_token="<unk>")
tokenizer.fit_on_texts(train_captions)
vocab_size = len(tokenizer.word_index) + 1
max_len = max(len(c.split()) for c in train_captions)
print(f"Vocab size: {vocab_size}, Max caption length: {max_len}")


def encode_captions(captions):
    return tokenizer.texts_to_sequences(captions)


train_seqs = encode_captions(train_captions)
test_seqs = encode_captions(test_captions)

# -------------------------------------------------------------------
# 3. Expand into (image, partial_caption) -> next_word training pairs
#    e.g. for "startseq a dress endseq" we create:
#       (img, [startseq])           -> a
#       (img, [startseq, a])        -> dress
#       (img, [startseq, a, dress]) -> endseq
# -------------------------------------------------------------------


def build_sequences(images, seqs):
    X_img, X_seq, y = [], [], []
    for img, seq in zip(images, seqs):
        for t in range(1, len(seq)):
            in_seq, out_word = seq[:t], seq[t]
            in_seq = pad_sequences([in_seq], maxlen=max_len)[0]
            X_img.append(img)
            X_seq.append(in_seq)
            y.append(out_word)
    return np.array(X_img), np.array(X_seq), to_categorical(y, num_classes=vocab_size)


X_img_train, X_seq_train, y_train_cap = build_sequences(train_images, train_seqs)
print("Training pairs:", X_img_train.shape[0])

# -------------------------------------------------------------------
# 4. Model: CNN encoder + LSTM decoder (classic "merge" architecture)
# -------------------------------------------------------------------
EMBED_DIM = 128

# ---- CNN branch: extracts a fixed-length feature vector from the image
# Fashion-MNIST images are small (28x28, grayscale), so a compact CNN is
# enough to get good features.
image_input = Input(shape=(28, 28, 1), name="image_input")
cnn = Conv2D(32, (3, 3), activation="relu", padding="same")(image_input)
cnn = MaxPooling2D()(cnn)
cnn = Conv2D(64, (3, 3), activation="relu", padding="same")(cnn)
cnn = MaxPooling2D()(cnn)
cnn = GlobalAveragePooling2D()(cnn)
image_features = Dense(EMBED_DIM, activation="relu")(cnn)
image_features = Dropout(0.3)(image_features)

# ---- LSTM branch: encodes the partial caption seen so far
seq_input = Input(shape=(max_len,), name="caption_input")
seq_embed = Embedding(vocab_size, EMBED_DIM, mask_zero=True)(seq_input)
seq_embed = Dropout(0.3)(seq_embed)
seq_features = LSTM(EMBED_DIM)(seq_embed)

# ---- Merge both branches and predict the next word
merged = Add()([image_features, seq_features])
merged = Dense(EMBED_DIM, activation="relu")(merged)
output = Dense(vocab_size, activation="softmax")(merged)

caption_model = Model(inputs=[image_input, seq_input], outputs=output)
caption_model.compile(optimizer="adam", loss="categorical_crossentropy",
                       metrics=["accuracy"])
caption_model.summary()

# -------------------------------------------------------------------
# 5. Train
# -------------------------------------------------------------------
history = caption_model.fit(
    [X_img_train, X_seq_train], y_train_cap,
    epochs=10, batch_size=64, validation_split=0.1, verbose=1
)

plt.figure(figsize=(6, 4))
plt.plot(history.history["loss"], label="train loss")
plt.plot(history.history["val_loss"], label="val loss")
plt.title("Caption model training loss")
plt.xlabel("Epoch")
plt.ylabel("Categorical cross-entropy")
plt.legend()
plt.tight_layout()
plt.savefig("caption_training_loss.png")
plt.close()

# -------------------------------------------------------------------
# 6. Inference: generate a caption word-by-word (greedy decoding)
# -------------------------------------------------------------------
index_to_word = {idx: word for word, idx in tokenizer.word_index.items()}


def generate_caption(image, max_length=max_len):
    in_text = "startseq"
    for _ in range(max_length):
        seq = tokenizer.texts_to_sequences([in_text])[0]
        seq = pad_sequences([seq], maxlen=max_length)
        pred = caption_model.predict([np.expand_dims(image, 0), seq], verbose=0)
        next_idx = int(np.argmax(pred))
        next_word = index_to_word.get(next_idx, "")
        if next_word == "" or next_word == "endseq":
            break
        in_text += " " + next_word
    return in_text.replace("startseq", "").strip()


# -------------------------------------------------------------------
# 7. Show a few sample predictions on unseen test images
# -------------------------------------------------------------------
fig, axes = plt.subplots(2, 3, figsize=(12, 8))
sample_indices = np.random.choice(len(test_images), 6, replace=False)
for ax, idx in zip(axes.ravel(), sample_indices):
    img = test_images[idx]
    true_label = class_names[int(y_test[test_idx[idx]])]
    pred_caption = generate_caption(img)
    ax.imshow(img.squeeze(), cmap="gray")
    ax.set_title(f"True: {true_label}\nGenerated: {pred_caption}", fontsize=9)
    ax.axis("off")
plt.tight_layout()
plt.savefig("caption_predictions.png")
plt.close()

print("\nSaved plots: caption_training_loss.png, caption_predictions.png")
print("Example generated caption:", generate_caption(test_images[0]))
```
